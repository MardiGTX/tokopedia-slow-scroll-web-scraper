```python
## Melakukan Import pandas, time, webdriver, beutifulsoup sebagai tools untuk Scraping website
import pandas as pd
import time

# Web Scraping
from selenium import webdriver
from bs4 import BeautifulSoup
```

```python
##Pembuatan Deklarasi 
namaproduk = []
harga = []
penjual = []
kotatoko = []
terjual = []
ratingp = []

##Melakukan pengambilan html website yang datanya mau di scrapping
driver = webdriver.Edge()
url = "https://www.tokopedia.com/search?st=&q=seblak&srp_component_id=02.01.00.00&srp_page_id=&srp_page_title=&navsource="
driver.get(url)

##Melakkan proses scroll website
for i in range(5): 
    driver.execute_script("window.scrollBy(0, 1000);") ##Slow Scroll buat prosesing load data tokopedia
    time.sleep(2)
    html = driver.page_source
    soup = BeautifulSoup(html, "html.parser")
        
    

    ## melakukan pembuatan proses boxes agar bisa memasukan data value yang kosong menjadi none secara paksa kedalam dataframe
    boxes = soup.find_all("a",{"class":"Ui5-B4CDAk4Cv-cjLm4o0g=="})
    ##melakukan proses scraping html website berdasarkan data yg ingin di ambil
    for box in boxes:
        elnama_produk = box.find("span", {"class":"+tnoqZhn89+NHUA43BpiJg=="})
        el_harga = box.find("div",{"class":"urMOIDHH7I0Iy1Dv2oFaNw=="})
        el_penjual = box.find("span",{"class":"si3CNdiG8AR0EaXvf6bFbQ=="})
        el_kotatoko = box.find("span",{"class":"gxi+fsEljOjqhjSKqjE+sw== flip"})
        el_terjual = box.find("span",{"class":"u6SfjDD2WiBlNW7zHmzRhQ=="})
        el_rating = box.find("span",{"class":"_2NfJxPu4JC-55aCJ8bEsyw=="})
        
        ##melakukan proses pemaksaan jika ada data yang kosong bisa di input menjadi none 
        try:
            namaproduk.append(elnama_produk.get_text())
        except:
            namaproduk.append(None)
        try:
            harga.append(el_harga.get_text())
        except:
            harga.append(None)
        try:
            penjual.append(el_penjual.get_text())
        except:
            penjual.append(None)
        try:
            kotatoko.append(el_kotatoko.get_text())
        except:
            kotatoko.append(None)
        try:
            terjual.append(el_terjual.get_text())
        except:
            terjual.append(None)
        try:
            ratingp.append(el_rating.get_text())
        except:
            ratingp.append(None)


##melakukan proses exit browser html
driver.quit()
##melakukan proses pembuatan dataframe
df = pd.DataFrame()

##melakukan proses pembuatan kolom kolom data frame dan penginputan data yang sudah didapat dari proses scrapping
df["namaproduk"] = namaproduk
df["harga"] = harga
df["penjual"] = penjual
df["kotatoko"] = kotatoko
df["terjual"] = terjual
df["rating"] = ratingp
##Melakukan Proses pembersihan data
df = df.drop_duplicates(subset=["namaproduk"]) 
df["harga"] = df["harga"].str.replace("Rp", "").str.replace(".", "").fillna("0") 
df["terjual"] = df["terjual"].str.replace("+", "").str.replace("terjual", "").str.replace("rb", "000").fillna("0")
df["rating"] = df["rating"].astype(float)
df["harga"] = df["harga"].astype(int)
df["terjual"] = df["terjual"].astype(int)

df.to_csv("scraping_seblak_tokped.csv")
```
